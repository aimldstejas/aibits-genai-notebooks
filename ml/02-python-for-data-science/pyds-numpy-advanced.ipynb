## Stacking Arrays

Stacking joins a sequence of arrays into one. `vstack` stacks row-wise (on top of each other), `hstack` stacks column-wise (side by side), and `column_stack` turns 1-D arrays into the columns of a 2-D array.

````python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("vstack:\n", np.vstack((a, b)))
print("hstack:", np.hstack((a, b)))
print("column_stack:\n", np.column_stack((a, b)))
````

`concatenate` is the general tool — pick the axis to join along. `axis=0` adds rows, `axis=1` adds columns.

````python
import numpy as np

p = np.array([[1, 2], [3, 4]])
q = np.array([[5, 6], [7, 8]])

print("concatenate axis=0:\n", np.concatenate((p, q), axis=0))
print("concatenate axis=1:\n", np.concatenate((p, q), axis=1))
````

`dstack` stacks depth-wise, producing a 3-D array — useful for combining, say, the R, G, B channels of an image.

````python
import numpy as np

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
d = np.dstack((a, b))
print("dstack shape:", d.shape)
print(d)
````

## Splitting Arrays

Splitting is the inverse of stacking. `split` divides a 1-D array into equal parts; `hsplit` and `vsplit` cut a 2-D array by columns and by rows respectively.

````python
import numpy as np

arr = np.arange(9)
print("split into 3:", np.split(arr, 3))

m = np.arange(1, 13).reshape(3, 4)
print("hsplit into 2 (columns):")
for part in np.hsplit(m, 2):
    print(part)

print("vsplit into 3 (rows):")
for part in np.vsplit(m, 3):
    print(part)
````

## Vectorised Operations

NumPy applies arithmetic to whole arrays at once, in fast C loops — no Python `for` loop needed. This "vectorisation" is why NumPy is the backbone of numerical Python.

````python
import numpy as np

a = np.arange(1, 6)
print("a * 2:", a * 2)
print("a ** 2:", a ** 2)
print("a + a:", a + a)
print("np.sqrt:", np.sqrt(np.array([1, 4, 9, 16])))
````

## Broadcasting Rules

Broadcasting lets NumPy operate on arrays of different shapes by stretching the smaller one. Comparing shapes from the **right**, two dimensions are compatible when they are **equal**, or **one of them is 1**. Here a (3,) row is added to every row of a (3,3) matrix.

````python
import numpy as np

A = np.array([[0, 0, 0], [10, 10, 10], [20, 20, 20]])
b = np.array([1, 2, 3])
print("A + b:\n", A + b)

col = np.array([[1], [2], [3]])   # (3,1)
row = np.array([10, 20, 30])      # (3,)
print("col + row:\n", col + row)
````

## Statistical Reductions

NumPy covers the everyday summary statistics; SciPy adds the geometric mean, harmonic mean, and mode. These are the numbers you compute on every new dataset.

````python
import numpy as np
from scipy.stats import gmean, hmean, mode

data = np.array([2, 4, 4, 4, 5, 5, 7, 9])
print("mean:", np.mean(data))
print("median:", np.median(data))
print("std:", np.std(data))
print("var:", np.var(data))
print("25th, 75th percentile:", np.percentile(data, [25, 75]))
print("geometric mean:", round(gmean(data), 4))
print("harmonic mean:", round(hmean(data), 4))
print("mode:", mode(data, keepdims=False).mode)
````

## Selecting & Ordering: where, sort, unique

Three reductions you'll use in nearly every preprocessing pipeline: `where` for conditional replacement, `sort` for ordering, and `unique` for distinct values.

````python
import numpy as np

a = np.array([3, 1, 4, 1, 5, 9, 2, 6])
print("where(a>3, a, 0):", np.where(a > 3, a, 0))   # keep if >3 else 0
print("sort:", np.sort(a))
print("argsort (index order):", np.argsort(a))
print("unique:", np.unique(a))
````

### A Small Matrix Workout

Generate a random 3×3 integer matrix (fixed seed for reproducibility), then take row sums, column maxima, and a flattened sorted view.

````python
import numpy as np

np.random.seed(0)
m = np.random.randint(1, 10, size=(3, 3))
print("matrix:\n", m)
print("row sums:", m.sum(axis=1))
print("column max:", m.max(axis=0))
print("flattened & sorted:", np.sort(m.flatten()))
````
