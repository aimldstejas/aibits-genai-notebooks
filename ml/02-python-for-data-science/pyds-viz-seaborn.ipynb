Seaborn sits on top of Matplotlib and makes *statistical* plots beautiful in a single line. It understands DataFrames directly, applies attractive defaults, and adds plot types — violin, swarm, pair — that would take dozens of Matplotlib lines.

## Why Seaborn?

- **Less code** — one function call vs. many Matplotlib commands, and no need to create a figure first.
- **DataFrame-aware** — pass `data=df` and reference columns by name; seaborn handles the rest.
- **Statistical by design** — built-in KDEs, confidence intervals, regression lines, and category-aware colouring (`hue`).
- **Attractive defaults** — sensible palettes and styling out of the box.

The examples below use two classic built-in datasets: `tips` (restaurant bills) and `iris` (flower measurements), loaded with `sns.load_dataset`.

## Histogram with KDE

`histplot` draws a histogram; `kde=True` overlays a smooth density curve — the distribution's shape at a glance.

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
sns.histplot(tips['total_bill'], kde=True, color='#7C3AED')
plt.title('Distribution of Total Bill')
plt.show()
````

## Scatter Plot with Categorical Colour

Passing `hue` colours the points by a category automatically — here, lunch vs dinner — turning a 2-variable scatter into a 3-variable view.

````python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill', y='tip', hue='time', data=tips)
plt.title('Tip vs Total Bill')
plt.show()
````

## Box Plot Across Categories

One box per category compares medians, spread, and outliers side by side.

````python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x='day', y='total_bill', data=tips, hue='day', palette='Purples', legend=False)
plt.title('Total Bill by Day')
plt.show()
````

## Correlation Heatmap

A heatmap of the correlation matrix shows every pairwise relationship at once, with `annot=True` printing the coefficients.

````python
import seaborn as sns
import matplotlib.pyplot as plt

corr = tips[['total_bill', 'tip', 'size']].corr()
sns.heatmap(corr, annot=True, cmap='PuOr', center=0, fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
````

A correlation heatmap is usually the single most information-dense chart in an EDA step — but two cautions matter every time you read one:

- It only shows **linear** (Pearson) relationships by default — a heatmap can show near-zero correlation for two variables that are, in fact, strongly related in a non-linear way (see the Inferential Statistics page).
- A correlation near ±1 between two *predictor* variables (not the target) signals **multicollinearity** — a problem for linear models specifically, addressed head-on on the Regularization page.

## Advanced: Violin, Swarm & Strip Plots

These three show the *distribution* of a numeric variable across categories, each with a different emphasis.

A **violin plot** combines a box plot with a mirrored KDE — wider sections mean more data at that value.

````python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset('iris')
sns.violinplot(x='species', y='petal_length', data=iris, hue='species',
               palette='Purples', legend=False)
plt.title('Petal Length by Species (Violin)')
plt.show()
````

A **swarm plot** shows every individual point, nudged sideways to avoid overlap — great for seeing the actual data density.

````python
sns.swarmplot(x='species', y='petal_length', data=iris, hue='species',
              palette=['#7C3AED', '#3B82F6', '#22C55E'], legend=False, size=3)
plt.title('Petal Length by Species (Swarm)')
plt.show()
````

A **strip plot** is similar but uses random jitter instead of a strict non-overlapping layout — lighter weight for larger data.

````python
sns.stripplot(x='species', y='petal_length', data=iris, hue='species',
              palette=['#7C3AED', '#3B82F6', '#22C55E'], legend=False, jitter=True)
plt.title('Petal Length by Species (Strip)')
plt.show()
````

## Multivariate: The Pair Plot

`pairplot` draws every numeric pair as a scatter and each variable's distribution on the diagonal, coloured by `hue` — a complete multivariate overview in one line.

````python
import seaborn as sns

sns.pairplot(iris, hue='species')
````

## Python Code: Every Chart Type on a Real Dataset (scikit-learn's Wine Dataset)

To make every row of the table above concrete, here's working code for all eight chart types run against a single real dataset — scikit-learn's built-in `load_wine`: 178 Italian wine samples, 13 chemical measurements, and a 3-class cultivar label. This is a genuinely non-Indian source dataset, so it's used as-is rather than reframed with Indian business context.

````python
from sklearn.datasets import load_wine
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = load_wine(as_frame=True)
df = data.frame
print(df.shape)
# (178, 14)  -- 178 samples, 13 chemical features + 1 target column
print(df['target'].value_counts().sort_index())
# 0    59
# 1    71
# 2    48
````

### Histogram — Shape of Alcohol Content

````python
plt.hist(df['alcohol'], bins=20)
plt.title('Distribution of Alcohol %')
# mean=13.00, std=0.81, min=11.03, max=14.83 -- roughly bell-shaped, mildly right-skewed
````

### Box Plot — Alcohol Spread by Cultivar Class

````python
sns.boxplot(x='target', y='alcohol', data=df)
# median alcohol -- class 0: 13.75, class 1: 12.29, class 2: 13.17
# class 0 wines run noticeably stronger than class 1
````

### Scatter Plot — Alcohol vs. Color Intensity

````python
plt.scatter(df['alcohol'], df['color_intensity'])
plt.xlabel('Alcohol %'); plt.ylabel('Color Intensity')
# Pearson r = 0.546 -- a moderate positive relationship, visible as an
# upward-drifting point cloud rather than a tight line
````

### Heatmap — Correlation Across Chemical Measurements

````python
subset = df[['alcohol', 'malic_acid', 'ash', 'color_intensity', 'proline']]
sns.heatmap(subset.corr(), annot=True, cmap='coolwarm')
# alcohol-proline r=0.64 (strongest pair shown), alcohol-malic_acid r=0.09 (near zero)
````

### Bar / Column Chart — Mean Alcohol per Cultivar Class

````python
sns.barplot(x='target', y='alcohol', data=df, estimator='mean')
# mean alcohol -- class 0: 13.74%, class 1: 12.28%, class 2: 13.15%
````

### Line Chart — Proline Trend Across Sample Order

````python
# The Wine dataset has no date/time column, so a 5-sample rolling average
# over row order stands in here purely to demonstrate the line-chart call.
df['proline'].rolling(5).mean().plot()
plt.ylabel('Proline (5-sample rolling mean)')
# first computed values (samples 5-9): 1103.0, 1180.0, 1228.0, 1250.0, 1163.0
````

### Pie Chart — Composition of the 3 Cultivar Classes

````python
counts = df['target'].value_counts().sort_index()
plt.pie(counts, labels=['Class 0','Class 1','Class 2'], autopct='%1.1f%%')
# 33.1% / 39.9% / 27.0% -- Class 1 is the largest cultivar group
````

### Count Plot — Frequency of Each Cultivar Class

````python
sns.countplot(x='target', data=df)
# counts -- class 0: 59, class 1: 71, class 2: 48
# same underlying numbers as the pie chart, shown here as raw bars instead of shares
````
