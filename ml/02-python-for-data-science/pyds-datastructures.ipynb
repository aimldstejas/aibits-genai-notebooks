## Built-in Data Structures

Four collection types cover almost every in-memory data-handling need. They differ along two axes: whether they are **ordered** and whether they are **mutable** (changeable after creation).

| Type | Ordered? | Mutable? | Duplicates? | Syntax |
|---|---|---|---|---|
| list | Yes | Yes | Yes | `[1, 2, 3]` |
| tuple | Yes | No | Yes | `(1, 2, 3)` |
| dict | Yes* | Yes | Keys: No | `{"a": 1}` |
| set | No | Yes | No | `{1, 2, 3}` |

*Dictionaries preserve insertion order since Python 3.7.

### Lists

Ordered, mutable, and the workhorse collection. Elements are reached by a zero-based index; negative indices count from the end.

````python
my_list = [1, 2, 3]
print(my_list)

my_list.append(4)          # add one item to the end
print(my_list)

my_list.extend([5, 6])     # add several items
print(my_list)

my_list.insert(0, 0)       # insert at a position
print(my_list)

my_list.remove(1)          # remove first matching value
print(my_list)

popped = my_list.pop(2)    # remove & return item at index 2
print(popped, my_list)

my_list.sort(reverse=True) # sort in place, descending
print(my_list)

my_list.reverse()          # reverse in place
print(my_list)
````

### Tuples

Ordered but **immutable** — perfect for fixed records (a coordinate, an RGB colour) and, because they're hashable, for use as dictionary keys.

````python
my_tuple = (1, 2, 3, 2, 4)
print(my_tuple[0], my_tuple[-3:])   # index + slice

print(my_tuple.count(2), my_tuple.index(3))   # count a value, find its position

a, b, c, d, e = my_tuple            # unpacking
print(a, e)
````

### Dictionaries

Key → value stores with fast lookup. The go-to structure for labelled records and configuration.

````python
person = {"name": "Isha", "age": 28, "city": "Surat"}
print(person["name"])

person["email"] = "isha@example.com"   # add a new pair
print(person)

print(person.get("age"), person.get("phone", "N/A"))   # safe lookup with default

for k, v in person.items():
    print(f"{k} -> {v}")
````

### Sets

Unordered collections of **unique** elements. Ideal for de-duplication and for mathematical set operations.

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a | b)     # union
print(a & b)     # intersection
print(a - b)     # difference
print(a ^ b)     # symmetric difference

nums = [1, 1, 2, 3, 3, 3, 4]
print(set(nums)) # de-duplicate a list
````

### List & Dictionary Comprehensions

A comprehension builds a collection in one readable line — shorter and usually faster than the equivalent `for` loop. The single most common idiom in day-to-day data prep.

````python
squares = [x ** 2 for x in range(10)]        # list comprehension
print(squares)

evens = [x for x in range(10) if x % 2 == 0]  # with a filter
print(evens)

sq_dict = {x: x ** 2 for x in range(1, 6)}    # dictionary comprehension
print(sq_dict)
````

### Three Short Data-Structure Tasks

Reverse a list, invert a dictionary (swap keys and values), and flatten a nested list — each a one-liner once you know the idiom.

````python
def reverse_list(lst):
    return lst[::-1]
print(reverse_list([1, 2, 3, 4, 5]))

def invert(d):
    return {v: k for k, v in d.items()}
print(invert({"a": 1, "b": 2, "c": 3}))

nested = [[1, 2], [3, 4], [5, 6]]
flat = [x for sub in nested for x in sub]
print(flat)
````

## Object-Oriented Programming

A **class** is a blueprint; an **object** is an instance built from it. OOP bundles data (attributes) and behaviour (methods) together, which keeps larger ML pipelines organised. It rests on four pillars:

### Classes & Objects

The `__init__` constructor runs when an object is created; `self` refers to the specific instance. A **class attribute** is shared by all instances; an **instance attribute** is per-object.

````python
class Dog:
    species = "Canis familiaris"        # class attribute (shared)

    def __init__(self, name, age):      # constructor
        self.name = name                # instance attribute
        self.age = age

    def bark(self):
        return f"{self.name} says Woof!"

d1 = Dog("Bruno", 3)
d2 = Dog("Kaia", 5)
print(d1.bark())
print(d2.name, d2.age, d2.species)
````

### Encapsulation

Bundle data with the methods that manage it, and mark internal state as private with a double underscore so it's changed only through controlled methods.

````python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance        # private (name-mangled)

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

acc = Account("Meera", 1000)
acc.deposit(500)
print(acc.owner, acc.get_balance())
````

### Inheritance

A child class inherits the parent's attributes and methods, then adds or overrides its own — avoiding duplicated code.

````python
class Bird:
    def __init__(self, name):
        self.name = name
    def intro(self):
        return f"{self.name} is a bird"

class Parrot(Bird):                     # Parrot inherits from Bird
    def fly(self):
        return f"{self.name} can fly"

p = Parrot("Mithu")
print(p.intro())                        # inherited method
print(p.fly())                          # own method
````

### Polymorphism

"Many forms": the same method name behaves differently per class, so you can treat different objects uniformly — call `.area()` on any shape without knowing which shape it is.

````python
class Shape:
    def area(self):
        return 0

class Circle(Shape):
    def __init__(self, r): self.r = r
    def area(self): return 3.14159 * self.r ** 2

class Square(Shape):
    def __init__(self, s): self.s = s
    def area(self): return self.s ** 2

for sh in [Circle(5), Square(4)]:
    print(f"{type(sh).__name__} area = {sh.area():.2f}")
````

### Special (Dunder) Methods

Methods named with double underscores let your objects work with built-in syntax. Defining `__add__` makes `+` work; `__str__` controls how `print` displays the object.

````python
class Vector:
    def __init__(self, x, y):
        self.x = x; self.y = y
    def __add__(self, other):           # enables the + operator
        return Vector(self.x + other.x, self.y + other.y)
    def __str__(self):                  # enables nice print()
        return f"Vector({self.x}, {self.y})"

v = Vector(2, 3) + Vector(4, 1)
print(v)
````

### Abstraction

An abstract base class defines *what* methods a subclass must provide without saying *how*. You can't instantiate it directly — it forces every concrete subclass to implement the required methods.

````python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass                            # no implementation here

class Car(Vehicle):
    def start(self):
        return "Car engine started"

c = Car()
print(c.start())

try:
    Vehicle()                           # can't instantiate an abstract class
except TypeError:
    print("Cannot instantiate abstract class")
````

class Wallet:
    def __init__(self, points=0):
        self.__points = points
    def add(self, n):
        self.__points += n
    def redeem(self, n):
        if n <= self.__points:
            self.__points -= n
            return True
        return False
    def balance(self):
        return self.__points
