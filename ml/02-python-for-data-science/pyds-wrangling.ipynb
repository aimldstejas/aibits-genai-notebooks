Real data arrives messy. Data wrangling — handling missing values, fixing types, removing duplicates, dropping useless columns, and scaling — is the unglamorous but essential step before any analysis or model.

Data wrangling (or "munging") reshapes raw data into a clean, analysis-ready form. Six operations cover the vast majority of the work:

## Handling Missing Values

Missing entries show up as `NaN`. `isnull().sum()` counts them per column. You then either fill them (`fillna`) — commonly with the column mean or median — or drop the affected rows (`dropna`).

````python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    'Product': ['Saree', 'Kurta', 'Shirt', 'Jeans', 'Dress', 'Scarf'],
    'Price':   [1200, 800, np.nan, 1500, np.nan, 300],
    'Units':   [5, np.nan, 8, 3, 6, np.nan],
})
print(df)
print("\nmissing per column:\n", df.isnull().sum())
````

Fill numeric gaps with a summary statistic — mean for roughly symmetric data, median when it's skewed or has outliers.

````python
df['Price'] = df['Price'].fillna(df['Price'].mean())
df['Units'] = df['Units'].fillna(df['Units'].median())
print(df)
````

When a row is too incomplete to trust, drop it. `dropna()` removes any row containing a `NaN`.

````python
import numpy as np, pandas as pd

df = pd.DataFrame({
    'Product': ['Saree', 'Kurta', 'Shirt', 'Jeans'],
    'Price':   [1200, np.nan, 900, 1500],
    'Units':   [5, 4, np.nan, 3],
})
print(df.dropna())
````

## Data Type Conversion

Columns often load as the wrong type — numbers stored as text, dates stored as strings. `astype` converts basic types; `pd.to_datetime` parses dates into proper datetime objects you can do arithmetic on.

````python
import pandas as pd

df = pd.DataFrame({
    'OrderID': ['101', '102', '103'],
    'Amount':  [1200.0, 800.0, 1500.0],
    'Date':    ['2024-01-05', '2024-01-06', '2024-01-07'],
})
print("before:\n", df.dtypes)

df['OrderID'] = df['OrderID'].astype(int)
df['Amount'] = df['Amount'].astype(int)
df['Date'] = pd.to_datetime(df['Date'])
print("\nafter:\n", df.dtypes)
````

## Duplicate Detection & Removal

`duplicated()` flags rows that repeat an earlier row; `drop_duplicates()` removes them. The `keep` argument controls which copy survives: `'first'` (default), `'last'`, or `False` to drop every duplicated row entirely.

````python
import pandas as pd

df = pd.DataFrame({
    'City': ['Mumbai', 'Surat', 'Mumbai', 'Pune', 'Surat'],
    'Sales': [200, 150, 200, 90, 150],
})
print("duplicated():\n", df.duplicated())
print("\nkeep first:\n", df.drop_duplicates())
print("\nkeep last:\n", df.drop_duplicates(keep='last'))
print("\ndrop all duplicated rows:\n", df.drop_duplicates(keep=False))
````

## Dropping Single-Valued Columns

A column where every row holds the same value carries no information for analysis or modelling. Find them with `nunique()` (unique-value count) and drop them.

````python
import pandas as pd

df = pd.DataFrame({
    'Product': ['A', 'B', 'C'],
    'Currency': ['INR', 'INR', 'INR'],   # same value everywhere
    'Price': [100, 200, 300],
})
nunq = df.nunique()
single = nunq[nunq == 1].index.tolist()
print("single-valued columns:", single)
print(df.drop(columns=single))
````

## Normalising & Standardising

Many algorithms expect features on a comparable scale. **Min-max normalisation** squeezes values into [0, 1]; **z-score standardisation** re-centres them to mean 0 and standard deviation 1.

````python
import pandas as pd

df = pd.DataFrame({'Score': [10, 20, 30, 40, 50]})
df['MinMax'] = (df['Score'] - df['Score'].min()) / (df['Score'].max() - df['Score'].min())
df['Zscore'] = (df['Score'] - df['Score'].mean()) / df['Score'].std()
print(df.round(3))
````

### A Full Clean-Up Pass

Put it together: count missing values, fill ages with the median, drop rows with no name, and remove duplicates.

````python
import numpy as np, pandas as pd

df = pd.DataFrame({
    'Name': ['Anaya', 'Rohan', None, 'Isha', 'Rohan'],
    'Age':  [28, np.nan, 35, 22, np.nan],
    'City': ['Mumbai', 'Surat', 'Pune', 'Mumbai', 'Surat'],
})
print("missing count:\n", df.isnull().sum())

df['Age'] = df['Age'].fillna(df['Age'].median())
df = df.dropna(subset=['Name']).drop_duplicates()
print("\ncleaned:\n", df)
````

```python
import numpy as np, pandas as pd
def clean(df):
    out = df.copy()
    single = [c for c in out.columns if out[c].nunique(dropna=False) == 1]
    out = out.drop(columns=single)
    for c in out.select_dtypes("number").columns:
        out[c] = out[c].fillna(out[c].median())
    return out.drop_duplicates()

```
