A chart reveals in seconds what a table hides in rows of numbers. Matplotlib is Python's foundational plotting library — every other viz tool (including seaborn) is built on it. Here: the anatomy of a figure, choosing the right chart, and building & customising the core plot types.

## Choosing the Right Chart

Before any code, pick the chart that matches your *question*. This is the single most important visualisation skill.

| Your question | Chart | Matplotlib call |
|---|---|---|
| Shape/spread of one numeric variable? | Histogram | `plt.hist(data)` |
| Trend over time? | Line | `plt.plot(x, y)` |
| Comparing a value across categories? | Bar | `plt.bar(cats, vals)` |
| Relationship between two numerics? | Scatter | `plt.scatter(x, y)` |
| Composition of a whole? | Pie | `plt.pie(vals, labels=...)` |
| Spread, median & outliers? | Box plot | `plt.boxplot(data)` |

## The Anatomy of a Matplotlib Figure

Three nested objects: the **Figure** is the whole canvas; it holds one or more **Axes** (an individual plot with its own data area); each Axes has two **Axis** objects (the x and y number lines). You size the figure with `figsize=(width, height)` in inches.

## Bar Chart — Comparing Categories

A bar chart compares a numeric value across categories. Every plot follows the same rhythm: create the figure, draw, then label the title and axes.

````python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [240, 300, 280, 360, 420, 390]

plt.figure(figsize=(7, 3.6))
plt.bar(months, sales, color='#7C3AED')
plt.title('Monthly Sales (Rupees thousands)')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
````

## Line Plot — Trends Over Time

A line plot connects data points in order — ideal for showing how a quantity changes over time. `marker='o'` highlights each data point.

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 3.6))
plt.plot(months, sales, marker='o', color='#3B82F6')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
````

## Pie Chart — Composition of a Whole

A pie chart shows each category's share of a total. `autopct` prints the percentage on each slice.

````python
import matplotlib.pyplot as plt

cities = ['Mumbai', 'Bengaluru', 'Hyderabad', 'Pune']
share = [40, 25, 20, 15]

plt.figure(figsize=(5, 4))
plt.pie(share, labels=cities, autopct='%1.0f%%', startangle=90)
plt.title('Sales Share by City')
plt.show()
````

## Customising: Colours, Legends, Labels & Titles

The difference between a rough plot and a clear one is customisation. To compare two series, plot both and add a **legend** so readers know which line is which.

````python
import matplotlib.pyplot as plt

product_a = [240, 300, 280, 360, 420, 390]
product_b = [180, 220, 260, 240, 300, 340]

plt.figure(figsize=(7, 3.6))
plt.plot(months, product_a, marker='o', color='#7C3AED', label='Product A')
plt.plot(months, product_b, marker='s', color='#F59E0B', label='Product B')
plt.title('Monthly Sales Comparison')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend(title='Product')       # the legend maps colour -> series
plt.show()
````

## Multiple Plots in One Figure — Subplots

`plt.subplots(rows, cols)` returns a figure and a grid of Axes, letting you place several charts side by side. Index the returned array to draw on each.

````python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(2, 2, figsize=(9, 6))
axs[0, 0].bar(months, sales, color='#7C3AED');      axs[0, 0].set_title('Bar')
axs[0, 1].plot(months, sales, marker='o', color='#3B82F6'); axs[0, 1].set_title('Line')
axs[1, 0].scatter(range(len(sales)), sales, color='#22C55E'); axs[1, 0].set_title('Scatter')
axs[1, 1].pie(share, labels=cities);                axs[1, 1].set_title('Pie')
fig.suptitle('Four Chart Types in a 2x2 Grid')
fig.tight_layout()
plt.show()
````
