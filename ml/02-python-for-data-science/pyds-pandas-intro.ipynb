Pandas adds *labels* to NumPy's arrays. The one-dimensional `Series` and the two-dimensional `DataFrame` are the structures you'll load, clean, filter, and analyse for the rest of this course.

A **Series** is a labelled 1-D array; a **DataFrame** is a labelled 2-D table — essentially a dictionary of Series sharing one index. Every column is a Series.

## Series

Create a Series from a list. Without an index, pandas assigns integer positions 0, 1, 2, … — just like a list, but you can attach your own labels instead.

````python
import pandas as pd

data = [10, 20, 30, 40, 50]
series = pd.Series(data)
print(series)
print("dtype:", series.dtype)
````

Supplying an `index` gives each value a meaningful label you can look up by.

````python
import pandas as pd

data = [10, 20, 30, 40, 50, 60]
index = ['a', 'b', 'c', 'x', 'y', 'z']
series = pd.Series(data, index=index)
print(series)
print("series['c']:", series['c'])
````

## DataFrame

The most common way to build a DataFrame is from a dictionary — each key becomes a column, each list becomes that column's values.

````python
import pandas as pd

data = {
    'City': ['Mumbai', 'Surat', 'Bengaluru'],
    'Total_Population': [340, 120, 280],
    'Literacy_Rate': [89.2, 85.4, 88.7],
}
df = pd.DataFrame(data)
print(df)
````

### Accessing Columns & Rows

Index a column by its label (returns a Series). Use `.iloc` for integer-position row access, and `.loc` with a boolean condition to select rows by value.

````python
import pandas as pd

data = {
    'City': ['Mumbai', 'Surat', 'Bengaluru'],
    'Total_Population': [340, 120, 280],
    'Literacy_Rate': [89.2, 85.4, 88.7],
}
df = pd.DataFrame(data)

print("column City:\n", df['City'])
print("first row (iloc[0]):\n", df.iloc[0])
print("rows where population == 340:\n", df.loc[df['Total_Population'] == 340])
````

### From a List of Records & Boolean Filtering

A list of dictionaries builds a DataFrame row by row. A comparison like `df['Age'] > 25` creates a boolean Series that filters the rows.

````python
import pandas as pd

data = [
    {'Name': 'Anaya', 'Age': 28, 'City': 'Mumbai'},
    {'Name': 'Rohan', 'Age': 34, 'City': 'Surat'},
    {'Name': 'Isha',  'Age': 22, 'City': 'Pune'},
]
df = pd.DataFrame(data)
print(df)
print("rows where Age > 25:\n", df.loc[df['Age'] > 25])
````

## Exploring a DataFrame

Once data is loaded, a handful of methods give you a fast overview. We'll use a synthetic 30-row employee table (fixed seed, so the numbers are reproducible) with three numeric and two categorical columns.

````python
import numpy as np
import pandas as pd

np.random.seed(42)
n = 30
df = pd.DataFrame({
    'Age': np.random.randint(22, 60, n),
    'Income': np.random.randint(30000, 120000, n),
    'Years_Employed': np.random.randint(1, 25, n),
    'Department': np.random.choice(['Engineering', 'Marketing', 'Finance'], n, p=[0.5, 0.3, 0.2]),
    'Location': np.random.choice(['Mumbai', 'Bengaluru', 'Hyderabad'], n, p=[0.5, 0.3, 0.2]),
})
print(df.head())          # first 5 rows
print("shape:", df.shape) # (rows, columns)
````

`dtypes` reports each column's type, and `describe()` summarises the numeric columns in one shot.

````python
print(df.dtypes)
print(df.describe().round(1))
````

For categorical columns, `value_counts()` tallies each category and `unique()` lists the distinct values.

````python
print(df['Department'].value_counts())
print("locations:", df['Location'].unique())
````
