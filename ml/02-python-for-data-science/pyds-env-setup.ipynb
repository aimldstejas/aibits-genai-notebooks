Before writing a line of ML code, set up a clean, reproducible Python environment. This page covers three ways to create isolated virtual environments — the built-in `venv`, the modern `uv`, and `conda` via Miniconda — plus registering a Jupyter kernel for each.

## Why a Separate Virtual Environment per Project?

A virtual environment is an isolated Python installation with its own packages. Creating a fresh one for *each* ML project is one of the highest-value habits you can build:

- **No dependency conflicts** — Project A can use pandas 1.5 while Project B uses pandas 2.2, without either breaking the other.
- **Reproducibility** — freeze the exact versions (`pip freeze > requirements.txt`) so the project runs identically on another machine months later.
- **A clean system Python** — you never pollute the OS-level Python, which the operating system itself depends on.
- **Easy cleanup** — delete one folder and the whole environment is gone, leaving no trace.
- **Shareable** — a teammate recreates your exact setup from `requirements.txt` in seconds.

> **💡 The golden rule**
>
> One project = one virtual environment = one Jupyter kernel. Never `pip install` into your global Python.

## Method 1 — `venv` (Built Into Python)

`venv` ships with Python 3.3+, so there is nothing extra to install. First make sure Python is available:

**Create** an environment named `myenv` in your project folder:

> **Run this one on your own computer (terminal or local Jupyter), not in Colab** (it is a shell command, or needs a desktop window, a running server, or keyboard input).

```python

```

**Activate** it — the command differs by operating system:

> **Run this one on your own computer (terminal or local Jupyter), not in Colab** (it is a shell command, or needs a desktop window, a running server, or keyboard input).

Your prompt now shows `(myenv)`. Install packages, and they land only inside this environment:

**Deactivate** when done, and **remove** by deleting the folder:

Terminal

> **Run this one on your own computer (terminal or local Jupyter), not in Colab** (it is a shell command, or needs a desktop window, a running server, or keyboard input).

```python

```

## Method 2 — `uv` (Modern & Fast)

`uv` is a Rust-based tool from Astral that replaces pip + virtualenv + pyenv and runs 10–100× faster. It can even install Python itself. **Install it first:**

macOS / Linux

> **Run this one on your own computer (terminal or local Jupyter), not in Colab** (it is a shell command, or needs a desktop window, a running server, or keyboard input).

```python
curl -LsSf https://astral.sh/uv/install.sh | sh
```

Windows PowerShell

> **Run this one on your own computer (terminal or local Jupyter), not in Colab** (it is a shell command, or needs a desktop window, a running server, or keyboard input).

```python
powershell -c "irm https://astral.sh/uv/install.ps1 | iex"
```

(Or, if you already have pip: `pip install uv`.) Then create an environment and install packages — note how fast it is:

Terminal

> **Run this one on your own computer (terminal or local Jupyter), not in Colab** (it is a shell command, or needs a desktop window, a running server, or keyboard input).

```python
# create a .venv in the current folder (optionally pin a Python version)
uv venv --python 3.12

# activate exactly like venv
source .venv/bin/activate        # macOS / Linux
# .venv\Scripts\activate         # Windows

# install packages (uv resolves + installs in milliseconds)
uv pip install numpy pandas scikit-learn matplotlib seaborn jupyter
```

> **💡 uv can manage Python too**
>
> `uv python install 3.12` downloads and manages Python versions for you — no separate installer needed. And `uv run script.py` auto-creates and uses the environment on the fly.

## Method 3 — `conda` via Miniconda

> **⚠ Use Miniconda, not Anaconda**
>
> **Miniconda** is the tiny installer — just `conda`, Python, and a few essentials (~100 MB) — versus the full Anaconda distribution (several GB of pre-installed packages you rarely need). Install what you actually use. Also install packages from the **conda-forge** channel, which is community-run and free of Anaconda's commercial-use licensing on the default channel.

**Install Miniconda.** On macOS/Linux, the official quick-install script:

Terminal

> **Run this one on your own computer (terminal or local Jupyter), not in Colab** (it is a shell command, or needs a desktop window, a running server, or keyboard input).

```python
# macOS / Linux
mkdir -p ~/miniconda3
# pick the right file for your OS: Linux-x86_64, MacOSX-arm64 (Apple Silicon), or MacOSX-x86_64
wget https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh -O ~/miniconda3/miniconda.sh
bash ~/miniconda3/miniconda.sh -b -u -p ~/miniconda3
rm ~/miniconda3/miniconda.sh
~/miniconda3/bin/conda init bash     # or: conda init zsh
```

On **Windows**, download the Miniconda installer `.exe` from `repo.anaconda.com/miniconda` and run it (tick "Add to PATH" if offered), then use the "Anaconda Prompt". Now **create and use** an environment:

Terminal

> **Run this one on your own computer (terminal or local Jupyter), not in Colab** (it is a shell command, or needs a desktop window, a running server, or keyboard input).

```python
# create an env with a specific Python, from the conda-forge channel
conda create -n myenv python=3.12 -c conda-forge

conda activate myenv                 # activate
conda install -c conda-forge numpy pandas scikit-learn matplotlib seaborn jupyter

conda deactivate                     # deactivate
conda env remove -n myenv            # remove entirely
```

## Registering a Jupyter Kernel

Creating an environment isn't enough — Jupyter needs to *see* it as a selectable kernel. Inside the activated environment, install `ipykernel` and register it for your user with the `--user` flag:

Terminal

> **Run this one on your own computer (terminal or local Jupyter), not in Colab** (it is a shell command, or needs a desktop window, a running server, or keyboard input).

```python
# (with your venv / uv / conda environment ACTIVATED)
pip install ipykernel        # or: conda install -c conda-forge ipykernel

# register this env as a Jupyter kernel for the current user
python -m ipykernel install --user --name myenv --display-name "Python (myenv)"
```

Now launch Jupyter and pick **"Python (myenv)"** from the kernel menu. The `--name` is the internal id; `--display-name` is what you see in the notebook. Manage kernels with:

Terminal

> **Run this one on your own computer (terminal or local Jupyter), not in Colab** (it is a shell command, or needs a desktop window, a running server, or keyboard input).

```python
jupyter kernelspec list                 # see all registered kernels
jupyter kernelspec uninstall myenv      # remove one
```

> **Which should you use?**
>
> **venv** — simplest, always available, perfect for beginners. **uv** — fastest and increasingly the modern default; great once you're comfortable. **conda/Miniconda** — best when you need non-Python dependencies (C/CUDA libraries) or precise scientific-stack builds. Any of them works for this course.
