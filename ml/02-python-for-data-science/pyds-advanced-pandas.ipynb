Precise selection is the difference between clumsy and fluent pandas. `.loc` selects by *label*, `.iloc` by *position* — master both, plus datetime indexing, `apply`, and `map`, and you can reshape any table to your will.

Every example below uses this small labelled table:

````python
import pandas as pd

df = pd.DataFrame({
    'Column1': [10, 20, 30, 40],
    'Column2': [45, 55, 65, 25],
    'Column3': ['a', 'b', 'c', 'd'],
}, index=['Row1', 'Row2', 'Row3', 'Row4'])
````

## Label-Based Selection with .loc

`.loc` uses the index and column *labels*. It handles single rows, specific row-column combinations, and boolean conditions.

````python
print("single row:\n", df.loc['Row2'])
print("\nrow + specific columns:\n", df.loc['Row3', ['Column1', 'Column3']])
print("\nconditional (Column2 > 50):\n", df.loc[df['Column2'] > 50])
````

## Position-Based Selection with .iloc

`.iloc` uses integer *positions*, exactly like list indexing — the end of a slice is excluded.

````python
print("third row (position 2):\n", df.iloc[2])
print("\nslice rows 1-2, cols 0-1:\n", df.iloc[1:3, 0:2])
print("\nsingle element [3, 1]:", df.iloc[3, 1])
````

## Datetime Indexing & Resampling

With a datetime index, you can select by date string and slice date ranges directly. `resample` then changes the time frequency — turning daily data into weekly or monthly summaries.

````python
import numpy as np, pandas as pd

np.random.seed(1)
dates = pd.date_range('2024-01-01', periods=100, freq='D')
df = pd.DataFrame({'Temp': np.random.randint(20, 35, 100)}, index=dates)

print("one date:\n", df.loc['2024-01-10'])
print("\ndate range:\n", df['2024-01-01':'2024-01-05'])
print("\nweekly mean (first 3 weeks):\n", df.resample('W').mean().head(3).round(2))
````

## Custom Functions with apply

`apply` runs a function along an axis of a DataFrame: `axis=0` processes each column, `axis=1` each row. It also accepts built-in vectorised functions like `np.sqrt`.

````python
import numpy as np, pandas as pd

df = pd.DataFrame({'A': [1, 2, 3], 'B': [10, 20, 30], 'C': [100, 200, 300]})

def value_range(col):
    return col.max() - col.min()

print("range per column (axis=0):\n", df.apply(value_range, axis=0))
print("\nrange per row (axis=1):\n", df.apply(value_range, axis=1))
print("\nnp.sqrt on every value:\n", df.apply(np.sqrt).round(2))
````

## Value Mapping with map

`map` transforms each value in a Series — via a dictionary (great for recoding categories) or a function.

````python
import pandas as pd

s = pd.Series([1, 2, 3, 4, 2])
print("dictionary mapping:\n", s.map({1: 'one', 2: 'two', 3: 'three', 4: 'four'}))
print("\nfunction mapping (x * 10):\n", s.map(lambda x: x * 10))
````

### Putting It Together

Label-slice a quarter range, filter high performers, and add a bonus column with `map`.

````python
import pandas as pd

df = pd.DataFrame({
    'Sales': [100, 250, 175, 300, 220],
    'Region': ['E', 'W', 'E', 'N', 'W'],
}, index=['Q1', 'Q2', 'Q3', 'Q4', 'Q5'])

print("Q2 to Q4 sales:\n", df.loc['Q2':'Q4', 'Sales'])
print("\nsales > 200:\n", df.loc[df['Sales'] > 200])

df['Bonus'] = df['Sales'].map(lambda x: x * 0.1)
print("\nwith bonus:\n", df)
````
