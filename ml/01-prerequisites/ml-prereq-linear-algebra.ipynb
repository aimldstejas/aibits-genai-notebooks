## Why Linear Algebra?

Every dataset this course touches — a Zomato ratings table, an Ames Housing spreadsheet, a Flipkart click-log — is, underneath the column names, a **matrix**: rows of numbers. Linear algebra is simply the mathematics of what you're allowed to do to grids of numbers: stretch them, rotate them, combine them, and — crucially for ML — find the small number of underlying directions that explain most of what's happening in a large number of columns. Where linear algebra stays purely linear (straight lines, flat planes, no curvature), calculus (next page) is what lets us climb down the slope of a cost function to actually *fit* a model. The two are partners: linear algebra gives data its shape, calculus gives models their learning rule.

## Vectors — The Basic Unit of Data

A **vector** is simply an ordered list of numbers. A single row of any dataset — one customer, one house, one transaction — is a vector once you strip away the column labels.

**Mehta Textiles weekly sales vector** (units sold): Cotton Saree, Silk Saree, Kurta 
**s** = [120, 85, 40]

### Vector Arithmetic

Vectors of the same length add and subtract element-by-element. Multiplying by a single number (a **scalar**) scales every element.

````python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)   # [5 7 9]  -- element-wise addition
print(a - b)   # [-3 -3 -3]
print(a * b)   # [4 10 18] -- element-wise product (NOT the dot product)
print(2.5 * a) # [2.5 5.  7.5] -- scalar multiplication
````

### The Dot Product

The **dot product** multiplies vectors element-wise and sums the result into a single number. It is, quite literally, how a weighted total gets computed — which is why it sits at the centre of linear regression, logistic regression, and every neuron-style weighted sum.

$$\mathbf{a}\cdot\mathbf{b} = \sum_i a_i b_i$$

````python
sales  = np.array([120, 85, 40])     # units: Cotton, Silk, Kurta
prices = np.array([1800, 4200, 900]) # ₹ per unit

revenue = sales.dot(prices)
print(revenue)  # 609000  -- total weekly revenue in ₹
````

609,000 is exactly (120×1800) + (85×4200) + (40×900) — the dot product is nothing more than "multiply corresponding entries, then add them all up." Every prediction ŷ = θ₀ + θ₁x₁ + θ₂x₂ + … in Linear Regression is a dot product between a feature vector and a weight vector, plus a bias term.

### Vector Norms — Measuring Length

A **norm** answers "how big is this vector?" Three norms recur constantly in ML, and each has a different personality:

| Norm | Formula | Where it shows up |
|---|---|---|
| L1 (Manhattan) | Σ\|aᵢ\| | Lasso regularization — encourages sparse (exactly-zero) weights |
| L2 (Euclidean) | √(Σaᵢ²) | Ridge regularization, KNN/K-Means distance, gradient magnitude |
| Max (L∞) | max(\|aᵢ\|) | Gradient clipping, worst-case error bounds |

````python
a = np.array([1, 2, 3])

print(np.linalg.norm(a, 1))        # 6.0    -- L1: |1|+|2|+|3|
print(np.linalg.norm(a, 2))        # 3.7417 -- L2: sqrt(1+4+9)
print(np.linalg.norm(a, np.inf))  # 3.0    -- max norm
````

> **📊 Prerequisite refresher**
>
> This is exactly the L1/L2 penalty term added to the cost function on the **Regularization** page (Ridge = squared L2 norm of weights; Lasso = L1 norm of weights) — the norm isn't a separate idea from regularization, it *is* the regularization term.

## Matrices — Grids of Vectors

Stack several vectors (rows) together and you have a **matrix** — precisely what a pandas DataFrame's underlying `.values` array is.

````python
# Mehta Textiles: 4 weeks x 3 products (Cotton, Silk, Kurta)
sales_matrix = np.array([
    [120, 85, 40],
    [135, 90, 55],
    [110,100, 60],
    [150, 95, 50]
])
prices = np.array([1800, 4200, 900])

weekly_revenue = sales_matrix.dot(prices)  # matrix-vector multiplication
print(weekly_revenue)
# [609000 670500 672000 714000]
print(weekly_revenue.sum())
# 2665500  -- total 4-week revenue in ₹
````

Matrix-vector multiplication is just the dot product applied row by row — this single operation is how every linear model (Linear Regression, Logistic Regression, Linear SVM) turns an entire dataset of feature vectors into a column of predictions in one shot: **ŷ** = **X**θ.

### Types of Matrices You'll Actually See

| Type | Property | ML relevance |
|---|---|---|
| Square | rows = columns | Required for a determinant/inverse to exist |
| Symmetric | A = Aᵀ | Every covariance matrix and kernel (Gram) matrix is symmetric |
| Diagonal | zeros off the main diagonal | A covariance matrix with uncorrelated features; scaling transforms |
| Identity (I) | 1s on diagonal, 0s elsewhere | AI = A — the "do nothing" transform; appears in Ridge's (XᵀX + λI) |
| Orthogonal | columns are unit vectors, mutually perpendicular; Qᵀ=Q⁻¹ | Rotation without distortion — exactly what PCA's eigenvectors and SVD's U, V are built from |

### Matrix Operations

````python
A = np.array([[1, 2], [3, 4]])

print(A.T)                     # transpose: [[1 3] [2 4]]
print(np.linalg.inv(A))       # inverse: [[-2.  1. ] [ 1.5 -0.5]]
print(np.linalg.det(A))       # determinant: -2.0
print(np.trace(A))            # trace (sum of diagonal): 5
print(np.linalg.matrix_rank(A)) # rank: 2 (full rank -- both rows independent)
````

| Operation | Intuition | Where it's used in this course |
|---|---|---|
| Transpose (Aᵀ) | Flip rows ↔ columns | The normal equation θ=(XᵀX)⁻¹Xᵀy on the Linear Regression page |
| Inverse (A⁻¹) | The matrix that "undoes" A: AA⁻¹=I | Solving the normal equation directly instead of iterating gradient descent |
| Determinant | How much A scales area/volume; 0 ⟺ not invertible | A zero (or near-zero) determinant of XᵀX is the tell-tale sign of multicollinearity |
| Rank | Number of genuinely independent rows/columns | Rank-deficient feature matrices cause the exact multicollinearity problem Ridge (next page but one) fixes |

> **⚠ Why Ridge adds λI instead of just using (XᵀX)⁻¹**
>
> When two features are highly correlated (e.g. "fabric weight in grams" and "fabric weight in kg" both present), XᵀX becomes singular or near-singular — its determinant approaches 0, and the inverse either doesn't exist or blows up numerically. Ridge Regression's (XᵀX + λI)⁻¹ adds a small positive value along the diagonal specifically to keep the matrix invertible. This is the linear-algebra reason Ridge exists, underneath the "shrinks coefficients" explanation given on the Regularization page.

## Systems of Linear Equations

Solving **Ax = b** for x is asking: "what combination of these directions (columns of A) reconstructs b?" This is exactly what the Normal Equation solves for regression — XᵀXθ = Xᵀy is a system of linear equations in θ, and θ = (XᵀX)⁻¹Xᵀy is just its closed-form solution.

````python
# Solve: 2x + y = 5, x + 3y = 10
A = np.array([[2, 1], [1, 3]])
b = np.array([5, 10])

x = np.linalg.solve(A, b)
print(x)  # [1. 3.]  -- i.e. x=1, y=3
````

## Matrix Decompositions

A decomposition rewrites a matrix as a product of simpler, more structured matrices — the linear-algebra equivalent of factoring a number. Three come up constantly in ML pipelines:

| Decomposition | Form | Used for |
|---|---|---|
| LU | A = PLU (lower × upper triangular) | Fast, stable solving of Ax=b under the hood in `np.linalg.solve` |
| QR | A = QR (orthogonal × upper triangular) | A numerically stabler way to solve the Normal Equation than direct inversion |
| Cholesky | A = LLᵀ (for symmetric positive-definite A) | Efficiently sampling from a Multivariate Gaussian — used on the Bayesian Approach & Gaussian Processes page |

````python
from scipy.linalg import cholesky

C = np.array([[4, 2], [2, 3]])   # symmetric positive-definite
L = cholesky(C, lower=True)
print(L)
# [[2.     0.    ]
#  [1.     1.4142]]
print(L.dot(L.T))  # reconstructs C exactly: [[4. 2.] [2. 3.]]
````

## Eigendecomposition — The Real Engine Behind PCA

An **eigenvector** of a matrix A is a direction that A doesn't rotate — it only stretches or shrinks it, by a factor called the **eigenvalue** (λ). Formally: **Av = λv**.

````python
B = np.array([[1, 2], [3, 4]])
vals, vecs = np.linalg.eig(B)

print(vals)   # [-0.3723  5.3723]  -- the two eigenvalues
print(vecs)   # each COLUMN is an eigenvector
# [[-0.8246 -0.416 ]
#  [ 0.5658 -0.9094]]

# Reconstruct B = V . diag(vals) . V^-1
V, Vinv = vecs, np.linalg.inv(vecs)
recon = V.dot(np.diag(vals)).dot(Vinv)
print(recon)  # [[1. 2.] [3. 4.]] -- exactly B, verified
````

### Why This Matters: Covariance → Eigenvectors → Principal Components

PCA's entire trick is to eigendecompose the **covariance matrix** of your features. The eigenvectors point along the directions of maximum spread in the data; the eigenvalues tell you exactly how much variance lies along each direction.

````python
# Cotton Saree units vs Silk Saree units, across the same 4 Mehta Textiles weeks
cotton = np.array([120, 135, 110, 150])
silk   = np.array([ 85,  90, 100,  95])

cov = np.cov(np.stack([cotton, silk]))
print(cov)
# [[306.25   -12.5  ]
#  [-12.5     41.667]]

vals, vecs = np.linalg.eig(cov)
print(vals)                      # [306.84  41.08]
print(vals / vals.sum())        # [0.882  0.118] -- explained variance ratio
````

88.2% of the joint variation in (Cotton, Silk) weekly sales lies along a single direction — that direction *is* Principal Component 1 on the PCA page. This is the exact eigendecomposition step that page's "Step 3" glosses over — now fully derived.

## Singular Value Decomposition (SVD)

Eigendecomposition only works cleanly on square matrices. **SVD** generalises the same idea — "find the important directions and how much they stretch" — to *any* rectangular matrix, which is why it's the workhorse behind PCA implementations, recommender systems, and the Kernel Methods page.

$$A = U\Sigma V^{\top}$$

````python
C = np.array([[1, 2], [3, 4], [5, 6]])  # 3x2, not square
U, s, VT = np.linalg.svd(C)

print(s)   # [9.5255  0.5143]  -- singular values, largest first

# Reconstruct C exactly
Sigma = np.zeros((3,2)); Sigma[:2,:2] = np.diag(s)
recon = U.dot(Sigma).dot(VT)
print(recon)  # [[1. 2.] [3. 4.] [5. 6.]] -- exact match
````

Keep only the largest singular value(s) and drop the rest, and U·Σ·Vᵀ becomes the best possible *low-rank approximation* of the original matrix — exactly the mechanism behind the Recommender Systems page's matrix factorization (compressing a huge, sparse user×item ratings matrix into two small dense ones) and an equivalent route to computing PCA without ever forming the covariance matrix directly.
