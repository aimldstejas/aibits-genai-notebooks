## 1. The hosted-API path
This notebook is written against an OpenAI-compatible chat-completions interface, which
most free-tier providers implement (Groq, Together, Fireworks, and others all offer one). Get
a free key from a provider of your choice, then set it below — **never paste a key directly
into a notebook cell you might share or commit; use Colab's Secrets panel** (key icon in the
left sidebar) and `userdata.get(...)`.

```python
!pip install -q openai transformers accelerate
```

```python
import os

try:
    from google.colab import userdata
    API_KEY = userdata.get('LLM_API_KEY')
    BASE_URL = userdata.get('LLM_BASE_URL')  # e.g. https://api.groq.com/openai/v1
except Exception:
    API_KEY = os.environ.get('LLM_API_KEY')
    BASE_URL = os.environ.get('LLM_BASE_URL')

hosted_available = bool(API_KEY and BASE_URL)
print('Hosted API configured:', hosted_available)
if not hosted_available:
    print('No key found — set LLM_API_KEY and LLM_BASE_URL in Colab Secrets to try this path.')
    print('The offline/local path below works regardless.')
```

```python
if hosted_available:
    from openai import OpenAI
    client = OpenAI(api_key=API_KEY, base_url=BASE_URL)
    response = client.chat.completions.create(
        model='llama-3.1-8b-instant',  # swap for whatever model your provider's free tier offers
        messages=[{'role': 'user', 'content': 'In one sentence, what is a transformer model?'}],
        max_tokens=60,
    )
    print('Hosted response:', response.choices[0].message.content)
    print('Tokens used:', response.usage.total_tokens)
else:
    print('Skipped — no hosted API configured.')
```

## 2. The offline / local path
This always works — no key, no network dependency beyond the one-time model download.
This is the fallback every lab in this course keeps available.

```python
from transformers import pipeline

gen = pipeline('text-generation', model='gpt2', device=-1)  # gpt2: small enough for CPU, no key needed
out = gen('In one sentence, a transformer model is', max_new_tokens=30, do_sample=True, top_p=0.9)
print('Local response:', out[0]['generated_text'])
```
