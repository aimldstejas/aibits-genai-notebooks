## 2. Basic tensor operations
The building block of everything in this course: a tensor, a matrix multiply, a gradient.

```python
import torch

x = torch.randn(4, 3, device=device, requires_grad=True)
w = torch.randn(3, 2, device=device, requires_grad=True)
y = x @ w
loss = y.pow(2).mean()
loss.backward()

print('x shape:', tuple(x.shape))
print('y = x @ w shape:', tuple(y.shape))
print('loss:', loss.item())
print('x.grad is not None:', x.grad is not None)
print('w.grad is not None:', w.grad is not None)
```

## 3. Load one row from each course dataset
Real datasets, per the course blueprint. Each loader tries the real source first and falls
back to a small synthetic stand-in if you have no internet connection right now — the offline
path always earns full marks in every graded lab in this course.

```python
# Chapter 2/3/4 dataset: UCI Default of Credit Card Clients (Meridian Bank credit-risk scenario)
import io
import urllib.request
import pandas as pd

UCI_CREDIT_URL = (
    'https://archive.ics.uci.edu/static/public/350/'
    'default+of+credit+card+clients.zip'
)

try:
    # fetch with a bounded timeout first - pd.read_excel(url) has no timeout of its own and
    # can hang the whole cell indefinitely on a stalled connection instead of falling back
    req = urllib.request.Request(
        'https://archive.ics.uci.edu/ml/machine-learning-databases/00350/'
        'default%20of%20credit%20card%20clients.xls',
        headers={'User-Agent': 'aibits-course-lab/1.0'},
    )
    with urllib.request.urlopen(req, timeout=30) as resp:
        raw = resp.read()
    credit_df = pd.read_excel(io.BytesIO(raw), header=1, index_col=0)
    print('Loaded UCI Credit Default from the source: ', credit_df.shape)
except Exception as e:
    print(f'Could not reach the UCI source ({e}). Using the offline fallback row.')
    credit_df = pd.DataFrame([{
        'LIMIT_BAL': 20000, 'SEX': 2, 'EDUCATION': 2, 'MARRIAGE': 1, 'AGE': 24,
        'BILL_AMT1': 3913, 'PAY_AMT1': 0, 'default payment next month': 1,
    }])

credit_df.head(1)
```

```python
# Chapter 6 dataset: UCI Metro Interstate Traffic Volume (Kestrel Logistics forecasting scenario)
try:
    req = urllib.request.Request(
        'https://archive.ics.uci.edu/ml/machine-learning-databases/00492/'
        'Metro_Interstate_Traffic_Volume.csv.gz',
        headers={'User-Agent': 'aibits-course-lab/1.0'},
    )
    with urllib.request.urlopen(req, timeout=30) as resp:
        raw = resp.read()
    traffic_df = pd.read_csv(io.BytesIO(raw), compression='gzip')
    print('Loaded Metro Interstate Traffic Volume from the source:', traffic_df.shape)
except Exception as e:
    print(f'Could not reach the UCI source ({e}). Using the offline fallback row.')
    traffic_df = pd.DataFrame([{
        'temp': 288.28, 'rain_1h': 0.0, 'clouds_all': 40,
        'date_time': '2012-10-02 09:00:00', 'traffic_volume': 5545,
    }])

traffic_df.head(1)
```

```python
# Chapter 7 dataset: Diabetes 130-US Hospitals (Lumina Health anomaly/embedding scenario)
try:
    import kagglehub
    path = kagglehub.dataset_download('brandao/diabetes')
    diabetes_df = pd.read_csv(f'{path}/diabetic_data.csv').head(1)
    print('Loaded Diabetes 130-US Hospitals from Kaggle:', diabetes_df.shape)
except Exception as e:
    print(f'Could not reach a live source ({e}). Using the offline fallback row.')
    diabetes_df = pd.DataFrame([{
        'race': 'Caucasian', 'gender': 'Female', 'age': '[70-80)',
        'time_in_hospital': 8, 'num_medications': 18, 'readmitted': '>30',
    }])

diabetes_df.head(1)
```

## 4. Plot one of them

```python
import matplotlib.pyplot as plt

cols = [c for c in ['LIMIT_BAL', 'AGE', 'BILL_AMT1'] if c in credit_df.columns]
credit_df[cols].head(min(len(credit_df), 200)).plot(subplots=True, figsize=(7, 4), title='Meridian Bank credit sample')
plt.tight_layout()
plt.show()
```
